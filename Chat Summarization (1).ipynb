```python
import re
from collections import Counter

print('✅ Libraries imported!')
```

```python
# Meeting conversation – change this text to any real conversation
meeting_text = """
Mohamed: Good morning everyone. Let's start the meeting.
We need to discuss the delay in the UI design task.
Mariam: I am still working on the design mockup. I need 2 more days to finish it.
Mohamed: Okay, we can extend the deadline to Friday.
Also, we need to fix the login bug before the demo.
Ahmed: I will fix the login bug by Thursday.
Mariam: I will submit the design mockup on Friday morning.
Mohamed: Great. Also everyone should review the project plan before Sunday meeting.
Rozana: I will prepare the meeting notes and send them to the team.
Mohamed: Perfect. Let's wrap up. See you all on Sunday.
"""
```

```python
# 1) Extract participants
def get_participants(text):
    participants = set()
    for line in text.strip().split('\n'):
        line = line.strip()
        if ':' in line:
            person = line.split(':')[0].strip()
            if person and len(person) < 20:
                participants.add(person)
    return list(participants)

print('✅ get_participants ready!')
```

```python
# 2) Meeting Summary
def summarize_meeting(text, top_n=3):
    lines = [l.strip() for l in text.strip().split('\n') if l.strip()]
    sentences = []
    for line in lines:
        if ':' in line:
            sentence = line.split(':', 1)[1].strip()
            if sentence:
                sentences.append(sentence)

    stop_words = {'i', 'we', 'the', 'a', 'an', 'is', 'are', 'was', 'to',
                  'and', 'or', 'but', 'in', 'on', 'at', 'by', 'for', 'of',
                  'it', 'let', 'all', 'you', 'me', 'my', 'will', 'can', 'also',
                  'good', 'great', 'okay', 'perfect', 'see', 'morning', 'everyone'}

    word_freq = Counter()
    for s in sentences:
        words = re.findall(r'\w+', s.lower())
        for w in words:
            if w not in stop_words and len(w) > 3:
                word_freq[w] += 1

    scored = []
    for s in sentences:
        words = re.findall(r'\w+', s.lower())
        score = sum(word_freq.get(w, 0) for w in words if w not in stop_words)
        scored.append((score, s))

    scored.sort(reverse=True)
    return [s for _, s in scored[:top_n]]

print('✅ summarize_meeting ready!')
```

```python
# 3) Action Items Extractor
def extract_action_items(text):
    action_keywords = ['will', 'going to', 'should', 'must', 'need to', 'have to']
    action_items = []

    for line in text.strip().split('\n'):
        line = line.strip()
        if not line or ':' not in line:
            continue
        line_lower = line.lower()
        if any(kw in line_lower for kw in action_keywords):
            person = line.split(':')[0].strip()
            action = line.split(':', 1)[1].strip()
            if person and len(person) < 20:
                action_items.append((person, action))

    return action_items

print('✅ extract_action_items ready!')
```

```python
# 4) Display results
print('=' * 50)
print('📋 MEETING ASSISTANT - TEAMIFY')
print('=' * 50)

participants = get_participants(meeting_text)
print(f'\n👥 Participants ({len(participants)}):')
for p in participants:
    print(f'  • {p}')

print('\n📝 Meeting Summary:')
summary = summarize_meeting(meeting_text)
for i, s in enumerate(summary, 1):
    print(f'  {i}. {s}')

print('\n✅ Action Items:')
actions = extract_action_items(meeting_text)
for person, action in actions:
    print(f'  - {person}: {action}')

lines = [l for l in meeting_text.strip().split('\n') if l.strip()]
print(f'\n📊 Stats:')
print(f'  • Total messages  : {len(lines)}')
print(f'  • Participants    : {len(participants)}')
print(f'  • Action items    : {len(actions)}')
print('=' * 50)
```
