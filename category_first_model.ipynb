```python
import json, pickle, os, warnings
import numpy as np
import pandas as pd
import torch
from sklearn.preprocessing import LabelEncoder, MultiLabelBinarizer
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, f1_score
from transformers import (
    DistilBertTokenizerFast, DistilBertForSequenceClassification,
    Trainer, TrainingArguments, pipeline,
)
from torch.utils.data import Dataset
warnings.filterwarnings("ignore")
print("✓ Imports done | Device:", "GPU" if torch.cuda.is_available() else "CPU")
```

```python
TASK_CSV    = "/content/task_dataset_v2.csv"          # ← upload to Colab
MEMBERS_CSV = "team_members_dataset.csv"  # ← upload to Colab

df = pd.read_csv(TASK_CSV,sep=';', header=0)
df["required_skills"] = df["required_skills"].apply(json.loads)
df["subtasks"]        = df["subtasks"].apply(json.loads)
df = df.dropna(subset=["task_text", "category", "difficulty"])
print(f"Task dataset   : {len(df):,} rows | {df['category'].nunique()} categories")
print(f"Category dist  :\n{df['category'].value_counts()}")
print(f"\nDifficulty dist:\n{df['difficulty'].value_counts()}")
```

# [CELL 4]
 TRAIN CATEGORY CLASSIFIER (DistilBERT)


```python
class TaskDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings; self.labels = labels
    def __getitem__(self, idx):
        item = {k: torch.tensor(v[idx]) for k, v in self.encodings.items()}
        item["labels"] = torch.tensor(self.labels[idx]); return item
    def __len__(self): return len(self.labels)

cat_le = LabelEncoder()
cat_labels = cat_le.fit_transform(df["category"])
print("Categories:", list(cat_le.classes_))

X_train_c, X_val_c, y_train_c, y_val_c = train_test_split(
    df["task_text"].tolist(), cat_labels, test_size=0.15, random_state=42, stratify=cat_labels
)

cat_tokenizer = DistilBertTokenizerFast.from_pretrained("distilbert-base-uncased")
cat_model     = DistilBertForSequenceClassification.from_pretrained(
    "distilbert-base-uncased", num_labels=len(cat_le.classes_)
)

enc_tr = cat_tokenizer(X_train_c, truncation=True, padding=True, max_length=64)
enc_vl = cat_tokenizer(X_val_c,   truncation=True, padding=True, max_length=64)
ds_tr  = TaskDataset(enc_tr, y_train_c)
ds_vl  = TaskDataset(enc_vl, y_val_c)

cat_args = TrainingArguments(
    output_dir="/content/drive/MyDrive/cat_model",
    num_train_epochs=3,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=64,
    warmup_steps=200,
    weight_decay=0.01,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    report_to="none",
)
cat_trainer = Trainer(model=cat_model, args=cat_args, train_dataset=ds_tr, eval_dataset=ds_vl)
cat_trainer.train()

preds = np.argmax(cat_trainer.predict(ds_vl).predictions, axis=1)
print(f"Category Accuracy: {accuracy_score(y_val_c, preds):.4f}")
print(classification_report(y_val_c, preds, target_names=cat_le.classes_))
```
